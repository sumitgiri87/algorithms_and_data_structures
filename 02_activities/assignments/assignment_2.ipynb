#### Starter Code for Question 2

- It is a Depth-first search (DFS) method. 

```python
# Definition for a binary tree node.
class TreeNode(object):
    def __init__(self, val = 0, left = None, right = None):
        self.val = val
        self.left = left
        self.right = right
from typing import List
# Defining the recursive function that lists all paths in a binary tree
def bt_path(root: TreeNode) -> List[List[int]]:
    all_paths=[]
    # Calling the recursive helper function that traverses ovel all paths
    all_paths_rec(root, all_paths,[])
    return all_paths

# Defing the helper function
def all_paths_rec(root, all_paths, current_path):
    if not root:
        return
    
    # adds the present node to the current path before executing the remaining function
    current_path.append(root.val)
    
    # When the recursive process reaches a leaf node 
    if not root.left and not root.right:
        all_paths.append(list(current_path))
    
    # When further branching possible, it checks the subtrees and calls the recursive helper function.
    else:
        all_paths_rec(root.left, all_paths, current_path)
        all_paths_rec(root.right, all_paths, current_path)
    
    # drops the present node from the current path in case in case of backtracking to the parent node.
    current_path.pop()
```

```python
# Example 1 test case
root1 = TreeNode(1)
root1.left = TreeNode(2)
root1.right = TreeNode(2)
root1.left.left = TreeNode(3)
root1.left.right = TreeNode(5)
root1.right.left = TreeNode(6)
root1.right.right = TreeNode(7)
bt_path(root1)
```

```python
# Example 2 test case
root2 = TreeNode(10)
root2.left = TreeNode(9)
root2.right = TreeNode(7)
root2.left.left = TreeNode(8)
bt_path(root2)
```

**Question One: First Duplicate in List:**
- *The function takes a list of integers and finds the first number that repeats. If more than one number repeats, it returns the one that appears first in the list. If there are no duplicates or the list is empty, it returns -1.* 


-   Create 1 new example that demonstrates you understand the problem. Trace/walkthrough 1 example that your partner made and explain it.


```python
# Input: nums = [4,5,1,2,5,9]
# Output: 5

# Input: nums = [1, -2, 2, 8, 4]
# Output: -1
```


-   Copy the solution your partner wrote. 


```python
from typing import List

def first_duplicate(nums: List[int]) -> int:
    seen = set()
    for num in nums:
        if num in seen:
            return num
        seen.add(num)
    return -1

# Test 1
nums = [3, 1, 4, 2, 5, 1, 6]
print("Input:", nums)
print("Output:", first_duplicate(nums))
print()

# Test 2
nums = [7, 8, 9, 10]
print("Input:", nums)
print("Output:", first_duplicate(nums))
print()

# Test 3
nums = [4, 5, 6, 4, 6]
print("Input:", nums)
print("Output:", first_duplicate(nums))
```


-   Explain why their solution works in your own words.


```python
# The function starts by creating an empty set to keep track of the numbers it has already seen. Then, it iterates through each number in the input list:
#   If a number is already present in the set of seen numbers, it immediately returns that number as the first duplicate.
#   If the number is not present, it gets added to the list.
# If the loop completes without finding any duplicates, the function returns -1 to indicate that all elements in the list are unique.
```


-   Explain the problem’s time and space complexity in your own words.


```python
# Time Complexity:

# The time complexity of this solution is O(n).
# The function iterates through all elements in the list once, which takes O(n) time. 
# For each element, it performs a membership check in a set, which is an O(1) operation on average because sets in Python are implemented using hash tables.
# Therefore, the overall time complexity is O(n) × O(1) = O(n).

# Space Complexity:

# The space complexity is O(n) in the worst case. This happens when all elements in the input list are unique, so the seen set grows to size n. Apart from this set, no additional significant memory is used.
```
