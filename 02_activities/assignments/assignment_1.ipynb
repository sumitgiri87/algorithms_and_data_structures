````python
# Examples 1:
# Input: s = "[]({[]}[()]){()}"
# Output: True

# Examples 2:
# Input: s = "{[()]{}}([]{(}))"
# Output: False
````


-   Code the solution to your assigned problem in Python (code chunk). Note: each problem can be solved more simply if you use an abstract data type that is suitable for that problem. Using that try to find the best time and space complexity solution!


````python
#This is the primary solution 
def is_valid_brackets(s: str) -> bool:
    stack=[]
    pair={')':'(',
          ']':'[',
          '}':'{'
          }
    for char in s:
        if char in pair:
            last_in_stack=stack.pop() if stack else '#'
            if pair[char] != last_in_stack:
                return False
        else:
            stack.append(char)
    return not stack

s_1="([]{})"
print(f'{s_1} is a valid bracket sequence: {is_valid_brackets(s_1)}')

s_2="([)]"
print(f'{s_2} is a valid bracket sequence: {is_valid_brackets(s_2)}')

s_3="()[]{}"
print(f'{s_3} is a valid bracket sequence: {is_valid_brackets(s_3)}')

s_4="[{]}"
print(f'{s_4} is a valid bracket sequence: {is_valid_brackets(s_4)}')

s_5="[]({[]}[()]){()}"
print(f'{s_5} is a valid bracket sequence: {is_valid_brackets(s_5)}')

s_6="{[()]{}}([]{(}))"
print(f'{s_6} is a valid bracket sequence: {is_valid_brackets(s_6)}')


````


-   Explain why your solution works


````python
# We first create a dictionary of key–value pairs where each closing bracket
# is mapped to its corresponding opening bracket.
#
# Next, we initialize an empty list to use as a stack.
#
# We then iterate through each character in the input string:
#   - If the character is a closing bracket, we check whether the top element
#     of the stack matches its corresponding opening bracket from the dictionary.
#     If it doesn’t match (or the stack is empty), the function returns False.
#   - Otherwise, we push it onto the stack.
#
# This process continues until all characters in the string have been processed.
#
# Finally, if the stack is empty, it means all brackets were properly matched
# and closed, so the function returns True. Otherwise, it returns False.
````


-   Explain the problem’s time and space complexity


````python
# Time Complexity:
# Iterating over all the characters in the string takes O(n) time.
# Within each iteration, the operations (pop, comparison, and append) each take O(1) time.
# Therefore, total time = O(n) × constant-time operations = O(n).

# Space Complexity:
# The stack only holds opening brackets that have not yet been matched.
# In the worst case, all characters could be opening brackets, so the stack size can grow up to n.
# Hence, the space complexity is O(n).
````
